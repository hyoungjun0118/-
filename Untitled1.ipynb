```python
try:
    from faker import Faker
except ImportError:
    !pip install faker
    from faker import Faker

import random

fake = Faker()

class User:
    def __init__(self, name=None):
        # 기본 정보 (과제 1)
        self.name = name if name else fake.name()
        self.email = fake.email()
        self.address = fake.address()
        self.job = fake.job()

        # 팔로우 정보 (과제 2)
        self.following = 0
        self.followers = 0

    # 팔로우 기능 (Tom이 Jerry 쫓아가기)
    def following_mode(self, other_user):
        self.following += 1
        other_user.followers += 1
        print(f"{self.name} → {other_user.name} 팔로우!")

    def __repr__(self):
        return f"{self.name} ({self.email})"


# =========================
# ✅ 과제 1: 1000명 생성
# =========================
users = [User() for _ in range(1000)]

print("생성된 유저 예시:")
for i in range(5):
    print(users[i])


# =========================
# ✅ 과제 2: Tom → Jerry 10번 팔로우
# =========================
tom = User("Tom")
jerry = User("Jerry")

print("\n[Tom이 Jerry를 쫓아가는 중]")
for _ in range(10):
    tom.following_mode(jerry)

print("\n결과")
print(f"Tom following: {tom.following}")
print(f"Jerry followers: {jerry.followers}")
```
